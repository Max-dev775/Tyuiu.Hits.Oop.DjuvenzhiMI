<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Описание задачи:
Создать базовый класс Order в C#, который будет представлять информацию о
заказах товаров или услуг. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.
Требования к базовому классу Order:
• Атрибуты: ID заказа (OrderId), Дата создания (CreationDate), Сумма заказа
(TotalAmount).
• Методы:
o CalculateTotal(): метод для расчета общей суммы заказа.
o AddItem(Item item): метод для добавления элемента в заказ.
o RemoveItem(Item item): метод для удаления элемента из заказа.
Требования к производным классам:
1. ОнлайнЗаказ (OnlineOrder): Должен содержать дополнительные атрибуты,
такие как Email клиента (CustomerEmail). Метод AddItem() должен быть
переопределен для добавления информации о способе доставки при
добавлении элемента.
2. ФизическийЗаказ (PhysicalOrder): Должен содержать дополнительные
атрибуты, такие как Адрес доставки (DeliveryAddress).
Метод RemoveItem() должен быть переопределен для добавления
информации о возврате товара при удалении элемента.
3. СпециализированныйЗаказ (SpecializedOrder) (если требуется третий класс):
Должен содержать дополнительные атрибуты, такие как Специальные
условия (SpecialConditions). Метод CalculateTotal() должен быть
переопределен для учета специальных условий при расчете общей суммы
заказа.

#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

<h2 style="color:DodgerBlue">Реализация:</h2>

```
// Класс Customer (Клиент)
public class Customer
{
    private string _fullName;
    private string _email;
    private string _phone;

    public string FullName
    {
        get => _fullName;
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Имя клиента не может быть пустым");
            _fullName = value;
        }
    }

    public string Email
    {
        get => _email;
        set
        {
            if (!string.IsNullOrWhiteSpace(value) && !value.Contains("@"))
                throw new ArgumentException("Некорректный email");
            _email = value;
        }
    }

    public string Phone
    {
        get => _phone;
        set => _phone = value;
    }

    public Customer(string fullName, string email, string phone)
    {
        FullName = fullName;
        Email = email;
        Phone = phone;
    }

    public override string ToString() => $"{FullName} ({Email})";
}

// Класс Item для товаров
public class Item
{
    private string _name;
    private decimal _price;
    private Order _order; // ссылка на заказ (взаимодействие объектов)

    public string Name
    {
        get => _name;
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Название товара не может быть пустым");
            _name = value;
        }
    }

    public decimal Price
    {
        get => _price;
        set
        {
            if (value < 0)
                throw new ArgumentException("Цена не может быть отрицательной");
            _price = value;
        }
    }

    // Свойство для взаимодействия с заказом
    public Order Order
    {
        get => _order;
        set => _order = value;
    }

    public Item(string name, decimal price)
    {
        Name = name;
        Price = price;
    }

    public override string ToString() => Name;
}

public class Order
{
    private int _orderId;
    private DateTime _creationDate;
    private decimal _totalAmount;
    protected List<Item> items = new List<Item>();
    private Customer _customer; // взаимодействие с клиентом

    public int OrderId
    {
        get => _orderId;
        set
        {
            if (value <= 0)
                throw new ArgumentException("ID заказа должен быть положительным");
            _orderId = value;
        }
    }

    public DateTime CreationDate
    {
        get => _creationDate;
        set => _creationDate = value;
    }

    public decimal TotalAmount
    {
        get => _totalAmount;
        protected set => _totalAmount = value;
    }

    // Взаимодействие с клиентом
    public Customer Customer
    {
        get => _customer;
        set => _customer = value;
    }

    public Order(int orderId, DateTime creationDate, Customer customer = null)
    {
        OrderId = orderId;
        CreationDate = creationDate;
        TotalAmount = 0;
        Customer = customer;
    }

    public virtual void CalculateTotal()
    {
        Console.WriteLine($"Общая сумма заказа #{OrderId}: {TotalAmount} руб.");
    }

    public virtual void AddItem(Item item)
    {
        items.Add(item);
        item.Order = this; // обратная связь: товар знает свой заказ
        TotalAmount += item.Price;
        Console.WriteLine($"Товар \"{item.Name}\" добавлен в заказ #{OrderId}");
    }

    public virtual void RemoveItem(Item item)
    {
        if (items.Contains(item))
        {
            items.Remove(item);
            item.Order = null;
            TotalAmount -= item.Price;
            Console.WriteLine($"Товар \"{item.Name}\" удален из заказа #{OrderId}");
        }
        else
        {
            Console.WriteLine($"Товар \"{item.Name}\" не найден в заказе #{OrderId}");
        }
    }

    public virtual void ShowOrder()
    {
        Console.WriteLine($"\nЗаказ #{OrderId}");
        Console.WriteLine($"Дата создания: {CreationDate}");
        Console.WriteLine($"Сумма: {TotalAmount} руб.");
        Console.WriteLine($"Товары: {string.Join(", ", items)}");
        if (Customer != null)
            Console.WriteLine($"Клиент: {Customer}");
    }
}

public class OnlineOrder : Order
{
    private string _customerEmail;

    public string CustomerEmail
    {
        get => _customerEmail;
        set
        {
            if (!string.IsNullOrWhiteSpace(value) && !value.Contains("@"))
                throw new ArgumentException("Некорректный email");
            _customerEmail = value;
        }
    }

    public OnlineOrder(int orderId, DateTime creationDate, string email, Customer customer = null)
        : base(orderId, creationDate, customer)
    {
        CustomerEmail = email;
    }

    public override void AddItem(Item item)
    {
        base.AddItem(item);
        Console.WriteLine($"Способ доставки: Курьерская доставка");
        Console.WriteLine($"Подтверждение отправлено на {CustomerEmail}");
    }

    public override void ShowOrder()
    {
        base.ShowOrder();
        Console.WriteLine($"Тип: Онлайн-заказ");
        Console.WriteLine($"Email клиента: {CustomerEmail}");
    }
}

public class PhysicalOrder : Order
{
    private string _deliveryAddress;

    public string DeliveryAddress
    {
        get => _deliveryAddress;
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Адрес доставки не может быть пустым");
            _deliveryAddress = value;
        }
    }

    public PhysicalOrder(int orderId, DateTime creationDate, string address, Customer customer = null)
        : base(orderId, creationDate, customer)
    {
        DeliveryAddress = address;
    }

    public override void RemoveItem(Item item)
    {
        base.RemoveItem(item);
        Console.WriteLine($"Информация о возврате товара \"{item.Name}\" отправлена по адресу: {DeliveryAddress}");
    }

    public override void ShowOrder()
    {
        base.ShowOrder();
        Console.WriteLine($"Тип: Физический заказ");
        Console.WriteLine($"Адрес доставки: {DeliveryAddress}");
    }
}

public class SpecializedOrder : Order
{
    private string _specialConditions;
    private decimal _discount;

    public string SpecialConditions
    {
        get => _specialConditions;
        set => _specialConditions = value;
    }

    public decimal Discount
    {
        get => _discount;
        set
        {
            if (value < 0 || value > 100)
                throw new ArgumentException("Скидка должна быть от 0 до 100%");
            _discount = value;
        }
    }

    public SpecializedOrder(int orderId, DateTime creationDate, string conditions, decimal discount = 0, Customer customer = null)
        : base(orderId, creationDate, customer)
    {
        SpecialConditions = conditions;
        Discount = discount;
    }

    public override void CalculateTotal()
    {
        decimal discountAmount = TotalAmount * Discount / 100;
        decimal finalTotal = TotalAmount - discountAmount;
        Console.WriteLine($"Общая сумма заказа #{OrderId}: {TotalAmount} руб.");
        Console.WriteLine($"Специальные условия: {SpecialConditions}");
        Console.WriteLine($"Скидка: {Discount}% ({discountAmount} руб.)");
        Console.WriteLine($"Итоговая сумма со скидкой: {finalTotal} руб.");
    }

    public override void ShowOrder()
    {
        base.ShowOrder();
        Console.WriteLine($"Тип: Специализированный заказ");
        Console.WriteLine($"Специальные условия: {SpecialConditions}");
        Console.WriteLine($"Скидка: {Discount}%");
    }
}

// Создаем клиента
Customer customer1 = new Customer("Иван Иванов", "ivan@mail.ru", "+7-900-123-45-67");

// Создаем товары
Item item1 = new Item("Смартфон", 50000);
Item item2 = new Item("Наушники", 5000);

// Онлайн-заказ
OnlineOrder onlineOrder = new OnlineOrder(1, DateTime.Now, "customer@mail.ru", customer1);
onlineOrder.AddItem(item1);
onlineOrder.AddItem(item2);
onlineOrder.CalculateTotal();
onlineOrder.ShowOrder();

//Взаимодействие объектов
Console.WriteLine("\nВзаимодействие объектов");
Console.WriteLine($"Товар \"{item1.Name}\" находится в заказе #{item1.Order?.OrderId}");
Console.WriteLine($"Клиент заказа #{onlineOrder.OrderId}: {onlineOrder.Customer?.FullName}");
```
