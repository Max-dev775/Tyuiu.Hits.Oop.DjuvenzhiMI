Описание задачи:
Создать базовый класс Order в C#, который будет представлять информацию о
заказах товаров или услуг. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.
Требования к базовому классу Order:
• Атрибуты: ID заказа (OrderId), Дата создания (CreationDate), Сумма заказа
(TotalAmount).
• Методы:
o CalculateTotal(): метод для расчета общей суммы заказа.
o AddItem(Item item): метод для добавления элемента в заказ.
o RemoveItem(Item item): метод для удаления элемента из заказа.
Требования к производным классам:
1. ОнлайнЗаказ (OnlineOrder): Должен содержать дополнительные атрибуты,
такие как Email клиента (CustomerEmail). Метод AddItem() должен быть
переопределен для добавления информации о способе доставки при
добавлении элемента.
2. ФизическийЗаказ (PhysicalOrder): Должен содержать дополнительные
атрибуты, такие как Адрес доставки (DeliveryAddress).
Метод RemoveItem() должен быть переопределен для добавления
информации о возврате товара при удалении элемента.
3. СпециализированныйЗаказ (SpecializedOrder) (если требуется третий класс):
Должен содержать дополнительные атрибуты, такие как Специальные
условия (SpecialConditions). Метод CalculateTotal() должен быть
переопределен для учета специальных условий при расчете общей суммы
заказа.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) исользуйтие в проекте коллекции, делегаты, события.

```
using System;
using System.Collections.Generic;
using System.Linq;

public interface ITrackable { void TrackStatus(); }
public interface ICancellable { void CancelOrder(); }

public delegate void OrderHandler(string message);

public class Customer
{
    private string _fullName;
    private string _email;
    private string _phone;
    private int _bonusPoints;

    public string FullName
    {
        get => _fullName;
        set { if (string.IsNullOrWhiteSpace(value)) throw new ArgumentException("Имя клиента не может быть пустым"); _fullName = value; }
    }

    public string Email
    {
        get => _email;
        set { if (!string.IsNullOrWhiteSpace(value) && !value.Contains("@")) throw new ArgumentException("Некорректный email"); _email = value; }
    }

    public string Phone { get => _phone; set => _phone = value; }

    public int BonusPoints
    {
        get => _bonusPoints;
        set { if (value < 0) throw new ArgumentException("Бонусные баллы не могут быть отрицательными"); _bonusPoints = value; }
    }

    public Customer(string fullName, string email, string phone)
    {
        FullName = fullName; Email = email; Phone = phone; BonusPoints = 0;
    }

    public void AddBonusPoints(int points)
    {
        BonusPoints += points;
        Console.WriteLine($"Клиенту {FullName} начислено {points} бонусных баллов. Всего: {BonusPoints}");
    }

    public override string ToString() => $"{FullName} ({Email})";
}

public class Item
{
    private string _name;
    private decimal _price;
    private Order _order;
    private int _quantity;

    public string Name
    {
        get => _name;
        set { if (string.IsNullOrWhiteSpace(value)) throw new ArgumentException("Название товара не может быть пустым"); _name = value; }
    }

    public decimal Price
    {
        get => _price;
        set { if (value < 0) throw new ArgumentException("Цена не может быть отрицательной"); _price = value; }
    }

    public Order Order { get => _order; set => _order = value; }

    public int Quantity
    {
        get => _quantity;
        set { if (value <= 0) throw new ArgumentException("Количество должно быть положительным"); _quantity = value; }
    }

    public Item(string name, decimal price) { Name = name; Price = price; Quantity = 1; }

    public decimal GetSubtotal() => Price * Quantity;

    public override string ToString() => Name;
}

public class Order
{
    private int _orderId;
    private DateTime _creationDate;
    private decimal _totalAmount;
    protected List<Item> items = new List<Item>();
    private Customer _customer;
    private string _status;
    private decimal _deliveryCost;
    private string _note;
    private int _priority;
    private DateTime _lastModified;

    private string _paymentMethod;
    private bool _isPaid;
    private string _managerName;

    public string OrderType { get; set; } = "Обычный";
    public string Region { get; set; } = "Не указан";
    public DateTime Deadline { get; set; }

    public event OrderHandler StatusChanged;

    public int OrderId
    {
        get => _orderId;
        set { if (value <= 0) throw new ArgumentException("ID заказа должен быть положительным"); _orderId = value; }
    }

    public DateTime CreationDate { get => _creationDate; set => _creationDate = value; }

    public decimal TotalAmount { get => _totalAmount; protected set => _totalAmount = value; }

    public Customer Customer { get => _customer; set => _customer = value; }

    public string Status { get => _status; set => _status = value; }

    public decimal DeliveryCost
    {
        get => _deliveryCost;
        set { if (value < 0) throw new ArgumentException("Стоимость доставки не может быть отрицательной"); _deliveryCost = value; }
    }

    public string Note { get => _note; set => _note = value; }

    public int Priority
    {
        get => _priority;
        set { if (value < 0) throw new ArgumentException("Приоритет не может быть отрицательным"); _priority = value; }
    }

    public DateTime LastModified { get => _lastModified; set => _lastModified = value; }

    public string PaymentMethod { get => _paymentMethod; set => _paymentMethod = value; }

    public bool IsPaid { get => _isPaid; set => _isPaid = value; }

    public string ManagerName { get => _managerName; set => _managerName = value; }

    public Order(int orderId, DateTime creationDate, Customer customer = null)
    {
        OrderId = orderId; CreationDate = creationDate; TotalAmount = 0;
        Customer = customer; Status = "Новый"; DeliveryCost = 0;
        Note = ""; Priority = 0; LastModified = creationDate;
        PaymentMethod = "Не указан"; IsPaid = false; ManagerName = "Не назначен";
        Deadline = creationDate.AddDays(7);
    }

    public virtual void CalculateTotal()
    {
        Console.WriteLine($"Общая сумма заказа #{OrderId}: {TotalAmount} руб.");
    }

    public virtual void CalculateTotal(decimal extraFee)
    {
        decimal finalTotal = TotalAmount + extraFee;
        Console.WriteLine($"Общая сумма заказа #{OrderId}: {TotalAmount} руб. + надбавка {extraFee} руб. = {finalTotal} руб.");
    }

    public virtual void AddItem(Item item)
    {
        items.Add(item); item.Order = this; TotalAmount += item.GetSubtotal();
        LastModified = DateTime.Now;
        Console.WriteLine($"Товар \"{item.Name}\" добавлен в заказ #{OrderId}");
    }

    public virtual void AddItem(Item item, int quantity)
    {
        item.Quantity = quantity;
        AddItem(item);
    }

    public virtual void RemoveItem(Item item)
    {
        if (items.Contains(item))
        {
            items.Remove(item); item.Order = null; TotalAmount -= item.GetSubtotal();
            LastModified = DateTime.Now;
            Console.WriteLine($"Товар \"{item.Name}\" удален из заказа #{OrderId}");
        }
        else Console.WriteLine($"Товар \"{item.Name}\" не найден в заказе #{OrderId}");
    }

    public virtual void ShowOrder()
    {
        Console.WriteLine($"\nЗаказ #{OrderId}");
        Console.WriteLine($"Дата создания: {CreationDate}");
        Console.WriteLine($"Статус: {Status}");
        Console.WriteLine($"Сумма: {TotalAmount} руб.");
        Console.WriteLine($"Стоимость доставки: {DeliveryCost} руб.");
        Console.WriteLine($"Товары: {string.Join(", ", items)}");
        if (Customer != null) Console.WriteLine($"Клиент: {Customer}");
    }

    public void UpdateStatus(string newStatus)
    {
        Status = newStatus;
        LastModified = DateTime.Now;
        StatusChanged?.Invoke($"Статус заказа #{OrderId} изменен на \"{newStatus}\"");
    }

    public virtual void ApplyDiscount(decimal percent)
    {
        if (percent < 0 || percent > 100) throw new ArgumentException("Скидка должна быть от 0 до 100%");
        decimal discountAmount = TotalAmount * percent / 100;
        TotalAmount -= discountAmount;
        LastModified = DateTime.Now;
        Console.WriteLine($"К заказу #{OrderId} применена скидка {percent}% ({discountAmount} руб.)");
    }

    public virtual string GetSummary() => $"Заказ #{OrderId} | {Status} | {TotalAmount} руб. | товаров: {items.Count}";

    public void MarkAsPaid()
    {
        IsPaid = true;
        LastModified = DateTime.Now;
        Console.WriteLine($"Заказ #{OrderId} оплачен, способ: {PaymentMethod}");
    }

    public void AssignManager(string manager)
    {
        ManagerName = manager;
        LastModified = DateTime.Now;
        Console.WriteLine($"Менеджер {manager} назначен на заказ #{OrderId}");
    }

    public int GetItemsCount() => items.Count;

    public void SetDeadline(DateTime deadline)
    {
        Deadline = deadline;
        Console.WriteLine($"Для заказа #{OrderId} установлен дедлайн: {deadline:d}");
    }

    public bool IsOverdue() => DateTime.Now > Deadline;
}

public class OnlineOrder : Order, ITrackable
{
    private string _customerEmail;
    private string _trackingNumber;
    private bool _emailNotifications;

    private string _website;
    private decimal _onlineDiscount;
    private bool _isExpress;

    public string PromoCode { get; set; } = "";
    public int BonusUsed { get; set; }
    public string DeviceType { get; set; } = "Неизвестно";

    public string CustomerEmail
    {
        get => _customerEmail;
        set { if (!string.IsNullOrWhiteSpace(value) && !value.Contains("@")) throw new ArgumentException("Некорректный email"); _customerEmail = value; }
    }

    public string TrackingNumber { get => _trackingNumber; set => _trackingNumber = value; }

    public bool EmailNotifications { get => _emailNotifications; set => _emailNotifications = value; }

    public string Website { get => _website; set => _website = value; }

    public decimal OnlineDiscount
    {
        get => _onlineDiscount;
        set { if (value < 0 || value > 100) throw new ArgumentException("Скидка должна быть от 0 до 100%"); _onlineDiscount = value; }
    }

    public bool IsExpress { get => _isExpress; set => _isExpress = value; }

    public OnlineOrder(int orderId, DateTime creationDate, string email, Customer customer = null)
        : base(orderId, creationDate, customer)
    {
        CustomerEmail = email; TrackingNumber = "Не присвоен"; EmailNotifications = true;
        Website = "shop.ru"; OnlineDiscount = 0; IsExpress = false;
    }

    public override void AddItem(Item item)
    {
        base.AddItem(item);
        Console.WriteLine($"Способ доставки: Курьерская доставка");
        Console.WriteLine($"Подтверждение отправлено на {CustomerEmail}");
    }

    public override void AddItem(Item item, int quantity)
    {
        base.AddItem(item, quantity);
        Console.WriteLine($"[Онлайн] Уведомление о {quantity} шт. \"{item.Name}\" отправлено на {CustomerEmail}");
    }

    public override void ShowOrder()
    {
        base.ShowOrder();
        Console.WriteLine($"Тип: Онлайн-заказ");
        Console.WriteLine($"Email клиента: {CustomerEmail}");
        Console.WriteLine($"Трек-номер: {TrackingNumber}");
    }

    public void TrackStatus()
    {
        Console.WriteLine($"Отслеживание онлайн-заказа #{OrderId}: статус \"{Status}\", трек-номер {TrackingNumber}");
    }

    public void SendNotification(string message)
    {
        Console.WriteLine($"[Email -> {CustomerEmail}] {message}");
    }

    public void ApplyOnlineDiscount()
    {
        if (OnlineDiscount <= 0) { Console.WriteLine("Онлайн-скидка не задана"); return; }
        ApplyDiscount(OnlineDiscount);
    }

    public void EnableExpress()
    {
        IsExpress = true;
        Console.WriteLine($"Заказ #{OrderId} помечен как экспресс-доставка");
    }

    public void ApplyPromoCode(string code)
    {
        PromoCode = code;
        Console.WriteLine($"К заказу #{OrderId} применен промокод: {code}");
    }
}

public class PhysicalOrder : Order, ICancellable
{
    private string _deliveryAddress;
    private string _courierName;
    private string _warehouse;

    private string _deliveryZone;
    private decimal _weight;
    private bool _requiresSignature;

    public string VehicleNumber { get; set; } = "Не назначен";
    public string PackagingType { get; set; } = "Стандартная";
    public int Floor { get; set; } = 1;

    public string DeliveryAddress
    {
        get => _deliveryAddress;
        set { if (string.IsNullOrWhiteSpace(value)) throw new ArgumentException("Адрес доставки не может быть пустым"); _deliveryAddress = value; }
    }

    public string CourierName { get => _courierName; set => _courierName = value; }

    public string Warehouse { get => _warehouse; set => _warehouse = value; }

    public string DeliveryZone { get => _deliveryZone; set => _deliveryZone = value; }

    public decimal Weight
    {
        get => _weight;
        set { if (value < 0) throw new ArgumentException("Вес не может быть отрицательным"); _weight = value; }
    }

    public bool RequiresSignature { get => _requiresSignature; set => _requiresSignature = value; }

    public PhysicalOrder(int orderId, DateTime creationDate, string address, Customer customer = null)
        : base(orderId, creationDate, customer)
    {
        DeliveryAddress = address; CourierName = "Не назначен"; Warehouse = "Основной склад";
        DeliveryZone = "Стандартная"; Weight = 0; RequiresSignature = false;
    }

    public override void RemoveItem(Item item)
    {
        base.RemoveItem(item);
        Console.WriteLine($"Информация о возврате товара \"{item.Name}\" отправлена по адресу: {DeliveryAddress}");
    }

    public override void ShowOrder()
    {
        base.ShowOrder();
        Console.WriteLine($"Тип: Физический заказ");
        Console.WriteLine($"Адрес доставки: {DeliveryAddress}");
        Console.WriteLine($"Курьер: {CourierName}");
    }

    void ICancellable.CancelOrder()
    {
        Status = "Отменен";
        LastModified = DateTime.Now;
        Console.WriteLine($"Физический заказ #{OrderId} отменен. Возврат по адресу: {DeliveryAddress}");
    }

    public void AssignCourier(string courier)
    {
        CourierName = courier;
        Console.WriteLine($"Курьер \"{courier}\" назначен на заказ #{OrderId}");
    }

    public void TransferToWarehouse(string warehouse)
    {
        Warehouse = warehouse;
        Console.WriteLine($"Заказ #{OrderId} перемещён на склад \"{warehouse}\"");
    }

    public void SetDeliveryZone(string zone)
    {
        DeliveryZone = zone;
        Console.WriteLine($"Зона доставки заказа #{OrderId}: {zone}");
    }

    public void RequireSignature()
    {
        RequiresSignature = true;
        Console.WriteLine($"Заказ #{OrderId} требует подписи при получении");
    }

    public void SetVehicle(string number)
    {
        VehicleNumber = number;
        Console.WriteLine($"Для заказа #{OrderId} назначен транспорт: {number}");
    }
}

public class SpecializedOrder : Order
{
    private string _specialConditions;
    private decimal _discount;
    private bool _isUrgent;
    private string _managerComment;

    private string _category;
    private bool _needsApproval;
    private decimal _minDeposit;

    public string ContractNumber { get; set; } = "";
    public bool IsConfidential { get; set; }
    public string ApprovalStage { get; set; } = "Не начато";

    public string SpecialConditions { get => _specialConditions; set => _specialConditions = value; }

    public decimal Discount
    {
        get => _discount;
        set { if (value < 0 || value > 100) throw new ArgumentException("Скидка должна быть от 0 до 100%"); _discount = value; }
    }

    public bool IsUrgent { get => _isUrgent; set => _isUrgent = value; }

    public string ManagerComment { get => _managerComment; set => _managerComment = value; }

    public string Category { get => _category; set => _category = value; }

    public bool NeedsApproval { get => _needsApproval; set => _needsApproval = value; }

    public decimal MinDeposit
    {
        get => _minDeposit;
        set { if (value < 0) throw new ArgumentException("Депозит не может быть отрицательным"); _minDeposit = value; }
    }

    public SpecializedOrder(int orderId, DateTime creationDate, string conditions, decimal discount = 0, Customer customer = null)
        : base(orderId, creationDate, customer)
    {
        SpecialConditions = conditions; Discount = discount; IsUrgent = false; ManagerComment = "";
        Category = "Общая"; NeedsApproval = false; MinDeposit = 0;
    }

    public override void CalculateTotal()
    {
        decimal discountAmount = TotalAmount * Discount / 100;
        decimal finalTotal = TotalAmount - discountAmount;
        if (IsUrgent) { finalTotal += 1000; Console.WriteLine("Срочный заказ: надбавка 1000 руб."); }
        Console.WriteLine($"Общая сумма заказа #{OrderId}: {TotalAmount} руб.");
        Console.WriteLine($"Специальные условия: {SpecialConditions}");
        Console.WriteLine($"Скидка: {Discount}% ({discountAmount} руб.)");
        Console.WriteLine($"Итоговая сумма со скидкой: {finalTotal} руб.");
    }

    public override void CalculateTotal(decimal extraFee)
    {
        decimal discountAmount = TotalAmount * Discount / 100;
        decimal finalTotal = TotalAmount - discountAmount + extraFee;
        if (IsUrgent) finalTotal += 1000;
        Console.WriteLine($"Заказ #{OrderId} (спец.): {TotalAmount} руб. - скидка {discountAmount} руб. + надбавка {extraFee} руб. = {finalTotal} руб.");
    }

    public override void ShowOrder()
    {
        base.ShowOrder();
        Console.WriteLine($"Тип: Специализированный заказ");
        Console.WriteLine($"Специальные условия: {SpecialConditions}");
        Console.WriteLine($"Скидка: {Discount}%");
        Console.WriteLine($"Срочный: {(IsUrgent ? "Да" : "Нет")}");
    }

    public void SendForApproval()
    {
        NeedsApproval = true;
        Console.WriteLine($"Заказ #{OrderId} отправлен на согласование, категория: {Category}");
    }

    public void SetDeposit(decimal amount)
    {
        MinDeposit = amount;
        Console.WriteLine($"Для заказа #{OrderId} установлен депозит: {amount} руб.");
    }

    public void SignContract(string number)
    {
        ContractNumber = number;
        Console.WriteLine($"Для заказа #{OrderId} подписан контракт: {number}");
    }
}

public class ExpressOnlineOrder : OnlineOrder
{
    private int _expressHours;

    public int ExpressHours
    {
        get => _expressHours;
        set { if (value <= 0) throw new ArgumentException("Срок экспресс-доставки должен быть положительным"); _expressHours = value; }
    }

    public ExpressOnlineOrder(int orderId, DateTime creationDate, string email, int expressHours, Customer customer = null)
        : base(orderId, creationDate, email, customer)
    {
        ExpressHours = expressHours;
    }

    public override void ShowOrder()
    {
        base.ShowOrder();
        Console.WriteLine($"Тип: Экспресс онлайн-заказ (доставка за {ExpressHours} ч.)");
    }
}

public class HybridOrder : Order, ITrackable, ICancellable
{
    public HybridOrder(int orderId, DateTime creationDate, Customer customer = null)
        : base(orderId, creationDate, customer) { }

    public void TrackStatus()
    {
        Console.WriteLine($"Отслеживание гибридного заказа #{OrderId}: статус \"{Status}\"");
    }

    public void CancelOrder()
    {
        Status = "Отменен";
        Console.WriteLine($"Гибридный заказ #{OrderId} отменен");
    }
}

public class OrderCollection<T> where T : Order
{
    private List<T> _orders = new List<T>();

    public void Add(T order) => _orders.Add(order);
    public void Remove(T order) => _orders.Remove(order);

    public void DisplayOrders()
    {
        foreach (var order in _orders)
            Console.WriteLine(order.GetSummary());
    }

    public List<T> FindByStatus(string status) => _orders.Where(o => o.Status == status).ToList();
    public T GetMostExpensive() => _orders.OrderByDescending(o => o.TotalAmount).FirstOrDefault();
}

public class OrderService
{
    private readonly Order _order;

    public OrderService(Order order)
    {
        _order = order;
    }

    public void ProcessOrder()
    {
        Console.WriteLine($"Сервис обрабатывает заказ #{_order.OrderId}, позиций: {_order.GetItemsCount()}");
        _order.CalculateTotal();
    }
}

Customer customer1 = new Customer("Иван Иванов", "ivan@mail.ru", "+7-900-123-45-67");
customer1.AddBonusPoints(500);

Item item1 = new Item("Смартфон", 50000) { Quantity = 1 };
Item item2 = new Item("Наушники", 5000) { Quantity = 2 };
Item item3 = new Item("Чехол", 500) { Quantity = 1 };

OnlineOrder onlineOrder = new OnlineOrder(1, DateTime.Now, "customer@mail.ru", customer1);
onlineOrder.AddItem(item1);
onlineOrder.TrackingNumber = "TRK-987654";
onlineOrder.ShowOrder();
onlineOrder.TrackStatus();

Console.WriteLine("");

ExpressOnlineOrder expressOrder = new ExpressOnlineOrder(2, DateTime.Now, "express@mail.ru", 3, customer1);
expressOrder.AddItem(item2);
expressOrder.ShowOrder();

Console.WriteLine("");

Console.WriteLine("Заказ #3");
HybridOrder hybridOrder = new HybridOrder(3, DateTime.Now, customer1);
hybridOrder.AddItem(item2);
hybridOrder.TrackStatus();
hybridOrder.CancelOrder();

Console.WriteLine("");

Console.WriteLine($"Товар \"{item1.Name}\" находится в заказе #{item1.Order?.OrderId}");
Console.WriteLine($"Клиент заказа #{onlineOrder.OrderId}: {onlineOrder.Customer?.FullName}");

Console.WriteLine("\nСпринт 3: перегрузки и новые методы");

onlineOrder.AddItem(item3, 3);
onlineOrder.CalculateTotal(500);

Console.WriteLine("");

PhysicalOrder physicalOrder = new PhysicalOrder(4, DateTime.Now, "г. Москва, ул. Ленина, 1", customer1);
physicalOrder.AddItem(item3);
physicalOrder.TransferToWarehouse("Северный склад");

Console.WriteLine("");

SpecializedOrder specialOrder = new SpecializedOrder(5, DateTime.Now, "Хрупкий товар", 15, customer1);
specialOrder.AddItem(item1);
specialOrder.CalculateTotal(300);

Console.WriteLine("");

OrderCollection<Order> orderCollection = new OrderCollection<Order>();
orderCollection.Add(onlineOrder);
orderCollection.Add(physicalOrder);
orderCollection.Add(specialOrder);
orderCollection.DisplayOrders();

Console.WriteLine("\nЗадание Sprint 4");

onlineOrder.PaymentMethod = "Карта";
onlineOrder.MarkAsPaid();
onlineOrder.AssignManager("Петров");
onlineOrder.EnableExpress();
onlineOrder.ApplyOnlineDiscount();

physicalOrder.Weight = 2.5m;
physicalOrder.SetDeliveryZone("Южная");
physicalOrder.RequireSignature();
ICancellable cancellable = physicalOrder;
cancellable.CancelOrder();

specialOrder.Category = "Хрупкое";
specialOrder.SendForApproval();
specialOrder.SetDeposit(5000);

Console.WriteLine("");
OrderService onlineService = new OrderService(onlineOrder);
onlineService.ProcessOrder();

OrderService physicalService = new OrderService(physicalOrder);
physicalService.ProcessOrder();

Console.WriteLine("\nСпринт 5: коллекции, делегаты, события");

onlineOrder.ApplyPromoCode("SALE10");
physicalOrder.SetVehicle("A123BC");
specialOrder.SignContract("CNT-2024-01");

onlineOrder.StatusChanged += OnStatusChanged;
onlineOrder.UpdateStatus("Отправлен");

var newOrders = orderCollection.FindByStatus("Новый");
Console.WriteLine($"Заказов со статусом \"Новый\": {newOrders.Count}");
Console.WriteLine($"Самый дорогой заказ: {orderCollection.GetMostExpensive().GetSummary()}");

void OnStatusChanged(string message)
{
    Console.WriteLine($"[Событие] {message}");
}
```
