```
using System;
using System.Collections.Generic;

public interface IInstallable { void Install(); }

public abstract class Component
{
    public string Name { get; set; }
    public decimal Price { get; set; }
    protected Component(string name, decimal price) { Name = name; Price = price; }
    public virtual string GetInfo() => $"{Name}, {Price} руб.";
    public override string ToString() => GetInfo();
}

public class CPU : Component, IInstallable
{
    public int Cores { get; set; }
    public CPU(string n, decimal p, int c) : base(n, p) => Cores = c;
    public override string GetInfo() => $"Процессор: {Name}, Ядер: {Cores}, {Price} руб.";
    public void Install() => Console.WriteLine($"[Установка] Процессор {Name}");
    public static bool operator ==(CPU a, CPU b) => a?.Name == b?.Name && a?.Price == b?.Price;
    public static bool operator !=(CPU a, CPU b) => !(a == b);
    public override bool Equals(object o) => o is CPU c && this == c;
    public override int GetHashCode() => HashCode.Combine(Name, Price);
}

public class RAM : Component, IInstallable
{
    public int SizeGB { get; set; }
    public RAM(string n, decimal p, int s) : base(n, p) => SizeGB = s;
    public override string GetInfo() => $"ОЗУ: {Name}, {SizeGB} ГБ, {Price} руб.";
    public void Install() => Console.WriteLine($"[Установка] ОЗУ {Name}");
    public static bool operator ==(RAM a, RAM b) => a?.Name == b?.Name && a?.Price == b?.Price;
    public static bool operator !=(RAM a, RAM b) => !(a == b);
    public override bool Equals(object o) => o is RAM r && this == r;
    public override int GetHashCode() => HashCode.Combine(Name, Price);
}

public class GPU : Component, IInstallable
{
    public int VramGB { get; set; }
    public GPU(string n, decimal p, int v) : base(n, p) => VramGB = v;
    public override string GetInfo() => $"Видеокарта: {Name}, {VramGB} ГБ, {Price} руб.";
    public void Install() => Console.WriteLine($"[Установка] Видеокарта {Name}");
    public static bool operator ==(GPU a, GPU b) => a?.Name == b?.Name && a?.Price == b?.Price;
    public static bool operator !=(GPU a, GPU b) => !(a == b);
    public override bool Equals(object o) => o is GPU g && this == g;
    public override int GetHashCode() => HashCode.Combine(Name, Price);
}

public class HDD : Component, IInstallable
{
    public int CapacityTB { get; set; }
    public HDD(string n, decimal p, int c) : base(n, p) => CapacityTB = c;
    public override string GetInfo() => $"HDD: {Name}, {CapacityTB} ТБ, {Price} руб.";
    public void Install() => Console.WriteLine($"[Установка] HDD {Name}");
    public static bool operator ==(HDD a, HDD b) => a?.Name == b?.Name && a?.Price == b?.Price;
    public static bool operator !=(HDD a, HDD b) => !(a == b);
    public override bool Equals(object o) => o is HDD h && this == h;
    public override int GetHashCode() => HashCode.Combine(Name, Price);
}

public class Computer
{
    private readonly List<Component> _components = new();

    public void AddComponent<T>(T c) where T : Component, IInstallable
    {
        c.Install();
        _components.Add(c);
    }

    public void RemoveComponent<T>(T c) where T : Component
        => Console.WriteLine(_components.Remove(c) ? $"[Удаление] {c.Name}" : $"[Удаление] {c.Name} не найден");

    public decimal GetTotalPrice() => _components.Sum(x => x.Price);

    public void Show()
    {
        Console.WriteLine("Конфигурация:");
        foreach (var x in _components) Console.WriteLine($" - {x}");
        Console.WriteLine($"Итого: {GetTotalPrice()} руб.\n");
    }
}

public static class ComputerFactory
{
    public static Computer CreateWithComponent<T>(T component) where T : Component, IInstallable
    {
        var pc = new Computer();
        pc.AddComponent(component);
        return pc;
    }

    public static Computer CreateOfficePC()
    {
        var pc = new Computer();
        pc.AddComponent(new CPU("Intel i3", 8000, 4));
        pc.AddComponent(new RAM("Kingston 8GB", 2000, 8));
        return pc;
    }

    public static Computer CreateGamingPC()
    {
        var pc = new Computer();
        pc.AddComponent(new CPU("Intel i7", 30000, 8));
        pc.AddComponent(new GPU("RTX 4070", 60000, 12));
        return pc;
    }
}

// Вывод

Console.WriteLine("1. Полиморфизм и перекрытие ToString");
var parts = new List<Component>
{
    new CPU("Intel i5", 15000, 6),
    new RAM("Kingston", 4000, 16),
    new GPU("RTX 4060", 35000, 8),
    new HDD("Seagate", 5000, 1)
};
foreach (var p in parts) Console.WriteLine(p);

Console.WriteLine("\n2. Операторы == и !=");
var r1 = new RAM("Kingston", 4000, 16);
var r2 = new RAM("Kingston", 4000, 16);
var r3 = new RAM("Corsair", 6000, 32);
Console.WriteLine($"r1 == r2: {r1 == r2}");
Console.WriteLine($"r1 != r2: {r1 != r2}");
Console.WriteLine($"r1 == r3: {r1 == r3}");
Console.WriteLine($"r1 != r3: {r1 != r3}");

Console.WriteLine("\n3. Полиморфизм через интерфейс IInstallable");
IInstallable[] installable = { new CPU("Intel i3", 8000, 4), new GPU("RTX 4060", 35000, 8), new HDD("WD", 5000, 2) };
foreach (var item in installable) item.Install();

Console.WriteLine("\n4. Сборка и изменение ПК (Add, Remove, GetTotalPrice)");
var pc = ComputerFactory.CreateWithComponent(new CPU("Intel i7", 30000, 8));
pc.AddComponent(new RAM("DDR5", 6000, 16));
pc.AddComponent(new GPU("RTX 4070", 60000, 12));
pc.AddComponent(new HDD("Samsung", 7000, 1));
pc.Show();
pc.RemoveComponent(new RAM("DDR5", 6000, 16));
pc.Show();

Console.WriteLine("5. Фабрика: офисная конфигурация");
ComputerFactory.CreateOfficePC().Show();

Console.WriteLine("6. Фабрика: игровая конфигурация");
ComputerFactory.CreateGamingPC().Show();
```
